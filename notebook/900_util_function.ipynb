```python
esg_data_dict = pd.read_json(esg_indicator_config_file_path)
esg_score_rules = pd.read_json(esg_score_rule_file_path)

with open(esg_company_file_path, 'r') as file:
    ec1 = json.load(file)
c1_df = pd.DataFrame(ec1['esg_context'])

esg_title_header = 'ESG report for ' + ec1['company_name'] + ' applied standard follow ' + ec1['esg_group'];

# print(esg_standard_df)
print(esg_title_header)
display(esg_data_dict)

display(esg_score_rules)
display(c1_df)
# display(c1_df[:1])

# def get_standard_by_key(key):
#     return esg_score_rules.query('metric_lable == @key') # filter(lambda o: o['metric_lable'] == key, esg_score_rules)[0]['standard_value']

# Function to get score based on metric_label and value
def get_score_by_metric(metric_label, value):
    print('===================metric_label:', metric_label)
    # print('esg_score_rules:', esg_score_rules)
    rules = esg_score_rules.loc[esg_score_rules['metric_label'] == metric_label] #.get(metric_label, [])
    print('======rules', rules)
    print('======value', value)
    for rule in rules:
        print('======rule checking', rule)
        if 'min_value' in rule and 'max_value' in rule:
            if rule['min_value'] <= value <= rule['max_value']:
                print('case1')
                return rule['score']
        elif 'max_value' in rule:
            if value <= rule['max_value']:
                print('case2')
                return rule['score']
        elif 'min_value' in rule:
            if value >= rule['min_value']:
                print('case3')
                return rule['score']
    print('case4')
    return None  # Or a default score if needed


# def get_indicator_score(input):
#     print(input)
#     rule = find_matched_rules(input['value'], get_standard_by_key(input['metric_lable']))
#     return rule['score']


# print(get_indicator_score(esg_data_dict[0]))
c1_df['score'] = c1_df.apply(lambda row: get_score_by_metric(row['metric_label'], row['value']), axis= 1)
display(c1_df)
# c1_with_score = c1_df

```
